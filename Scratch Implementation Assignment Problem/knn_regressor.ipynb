```python
import numpy as np
import pandas as pd
```

```python
class KNNRegressor:
    def __init__(self,k=3):
        self.k=k
    def fit(self,X,y):
        self.X_train=X
        self.y_train=y
    def _eucilian_dist(self,x1,x2):
        return np.sqrt(np.sum((x1-x2)**2))
    def _predict_one(self,X):
        distances=[self._eucilian_dist(X,x_train) for x_train in self.X_train]
        knn_indices=np.argsort(distances)[:self.k]
        knn_classes=[self.y_train[i] for i in knn_indices]
        average_classes=np.mean(knn_classes)
        return average_classes
    def predict(self,X):
        y_predict=[self._predict_one(x) for x in X]
        return np.array(y_predict)
```

```python
#Training Data
X_train = np.array([[1], [2], [3], [4], [5]])
y_train = np.array([2, 4, 6, 8, 10])
#Testing Data
X_test = np.array([[1.5], [3.5], [6]])
model=KNNRegressor()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
print(y_pred)
```

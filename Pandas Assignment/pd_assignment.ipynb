```python
import pandas as pd
```

```python
data=pd.read_csv("IRIS.csv")
data1=data.copy()
```

```python
#Q1
#A
ten_rows=data1.head(10)
data1.shape
data1.dtypes
```

```python
#B
typea="Iris-virginica"
selected_data1=data1.query("petal_length>4.5 and species==@typea")
selected_data1
```

```python
#C
data1.groupby("species").agg(
    avg_sepal_length=("sepal_length","mean"),
    max_petal_width=("petal_width","max"),
    std_sepal_width=("sepal_width","std")
)
```

```python
#D
data1.assign(petal_ratio=data["petal_length"]/data["petal_width"])
data1.groupby("species")["petal_ratio"].mean()
```

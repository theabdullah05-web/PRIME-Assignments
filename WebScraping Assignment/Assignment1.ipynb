```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import lxml
```

```python
page_count=1
while(True):
    res=requests.get(f"https://quotes.toscrape.com/page/{page_count}")
    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")
    if(not quotes):
        print("No content found on this page...")
        break;
    with open(f"scraped_data/quote{page_count}.html","w",encoding="utf-8") as f:
        f.write(res.text)
    print(f"downloaded data from page{page_count}")
    page_count=page_count+1
```

```python
count=1
quote_list=[]
while(True):
    try:
        with open(f"scraped_data/quote{count}.html","r",encoding="utf-8") as f:
            data=f.read()
        soup=BeautifulSoup(data,"lxml")
        all_quotes=soup.select(".quote")
        for quote in all_quotes:
            all_tags=[]
            for tag in quote.select(".tags .tag"):
                all_tags.append(tag.get_text())
            if "life" in all_tags:
                quote_list.append({"Author":quote.select_one(".author").get_text(),"Quote":quote.select_one(".text").get_text(),"Page":count,"Tags":all_tags})
        count=count+1
    except FileNotFoundError:
        break;
```

```python
df=pd.DataFrame(quote_list)
```

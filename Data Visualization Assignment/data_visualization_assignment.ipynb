```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Q1
np.random.seed(0)
scores=np.random.normal(70,10,200)
plt.hist(scores,edgecolor="black",bins=20)
plt.axvline(
    x=scores.mean(),
    color="yellow"
)
```

```python
#Q2
penguins=sns.load_dataset("penguins")
penguins=penguins.dropna()
fig,ax=plt.subplots()
sns.scatterplot(
    data=penguins,
    x="body_mass_g",
    y="flipper_length_mm",
    hue="species",
    ax=ax
)
ax.set_title("Penguins")
```

```python
#Q3
tips=sns.load_dataset("tips")
fig,ax=plt.subplots()
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax
)
fig.tight_layout()
```

```python
#Q4
tips=sns.load_dataset("tips")
fig,ax=plt.subplots()
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    ax=ax
)
ax.set_xlabel("Total Bill in USD")
ax.set_ylabel("Tip in USD")
fig.suptitle("Total Bill vs Tip")
```

```python
tips=sns.load_dataset("tips")
tips=tips.pivot_table(index="day",columns="time",values="tip",observed=True)
tips=tips.fillna(0)
fig,ax=plt.subplots()
sns.heatmap(
    tips,
    cmap="coolwarm",
    annot=True,
)
fig
fig.tight_layout()
```
